```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Zomato_Dataset.csv")

print(df.head())
```

```python
print(df.shape)

print(df.columns)

print(df.info())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Rating"], bins=10)

plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Restaurants")

plt.show()
```

```python
top_cities = df["City"].value_counts().head(10)

top_cities.plot(kind="bar", figsize=(10, 5))

plt.title("Top 10 Cities with Most Restaurants")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
df["Price_range"].value_counts().sort_index().plot(
    kind="bar", figsize=(8, 5)
)

plt.title("Distribution of Restaurant Price Ranges")
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")

plt.show()
```

```python
df["Has_Online_delivery"].value_counts().plot(
    kind="bar", figsize=(8, 5)
)

plt.title("Restaurants Offering Online Delivery")
plt.xlabel("Online Delivery")
plt.ylabel("Number of Restaurants")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Votes"], df["Rating"], alpha=0.5)

plt.title("Votes vs Restaurant Ratings")
plt.xlabel("Votes")
plt.ylabel("Rating")

plt.show()
```

```python
sns.boxplot(x="Price_range", y="Rating", data=df)

plt.title("Restaurant Ratings by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Rating")

plt.show()
```
